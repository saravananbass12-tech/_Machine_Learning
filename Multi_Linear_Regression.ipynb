```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
data=pd.read_csv("StudentsPerformance.csv")
data
```

```python
x=data[['reading score','writing score']]
x
```

```python
y=data[['math score']]
y
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
reading=int(input("enter your reading score:"))
writing=int(input("enter your writing score:"))
print("your math score is:",model.predict([[reading,writing]]))
```

```python
x_test=x
y_test=y
y_pred=model.predict(x_test)
accuracy=r2_score(y_test,y_pred)
print("r2_score:",accuracy)
print("r2_score:",accuracy*100)
```

```python
import matplotlib.pyplot as plt
fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')
ax.scatter(data['reading score'],data['writing score'],data['math score'],color='#FF007F')
plt.title("students performance")
ax.set_xlabel("reading score")
ax.set_ylabel("writing score")
ax.set_zlabel("math score")
plt.show()
```
