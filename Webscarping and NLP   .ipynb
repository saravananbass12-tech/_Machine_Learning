```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.w3schools.com/w3css/w3css_tables.asp"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")
print(f"Response status code: {response.status_code}")
print(f"HTML snippet (first 500 chars): {response.text[:500]}")
```

```python
table = soup.find("table")
rows = table.find_all("tr") if table else []

data = []
for row in rows:
    cols = row.find_all(["td", "th"])
    cols = [c.text.strip() for c in cols]
    data.append(cols)
```

```python
df=pd.DataFrame(data[1:], columns=data[0])
df
```

```python
import nltk

# Ensure required tokenizers and datasets are downloaded
nltk.download('punkt_tab')

# Get user input as a string, not an integer
sentence = input("Enter the sentence: ")

# Tokenize the sentence
tokens = nltk.word_tokenize(sentence.lower())

# Lemmatize the tokens
lemmatizer = nltk.WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in tokens]

print("Lemmas:", lemmas)

# Define opinion words
positive_words = ["good", "nice", "excellent", "happy", "love"]
negative_words = ["bad", "worst", "sad", "poor", "hate"]

# Count occurrences
positive_count = sum(1 for word in lemmas if word in positive_words)
negative_count = sum(1 for word in lemmas if word in negative_words)

print("Positive count:", positive_count)
print("Negative count:", negative_count)
```
